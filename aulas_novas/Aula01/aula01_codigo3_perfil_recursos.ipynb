## 1. A tabela de recursos

Escala de suporte, atribuída pelo professor a partir da documentação oficial de cada linguagem:

| Nota | Significado |
|---|---|
| **0** | ausente ou só por biblioteca externa com custo alto |
| **1** | possível, mas não idiomático — dá trabalho |
| **2** | suportado e idiomático |
| **3** | é o modo padrão de escrever nessa linguagem |

Estes valores são um **instrumento didático**, não uma medida objetiva. Discordar deles com
argumento é exatamente o exercício que a aula pede.


```python
import pandas as pd

RECURSOS = [
    "Funções de 1ª classe", "Closures", "Imutabilidade por padrão",
    "Tipos algébricos", "Pattern matching", "Avaliação preguiçosa",
    "Inferência de tipos", "Despacho dinâmico", "Herança de classes",
    "Unificação / lógica", "Atores ou canais", "Macros / metaprogramação",
]

dados = {
    "Python":  [3, 3, 0, 1, 2, 1, 1, 3, 3, 0, 1, 2],
    "Java":    [2, 2, 0, 1, 2, 0, 1, 3, 3, 0, 1, 1],
    "Haskell": [3, 3, 3, 3, 3, 3, 3, 1, 0, 0, 1, 2],
    "Elixir":  [3, 3, 3, 2, 3, 1, 0, 2, 0, 0, 3, 3],
    "Racket":  [3, 3, 2, 2, 2, 1, 1, 2, 1, 1, 3, 3],
    "Prolog":  [1, 0, 3, 1, 3, 1, 0, 0, 0, 3, 0, 2],
    "Go":      [2, 2, 0, 0, 0, 0, 1, 2, 0, 0, 3, 0],
}

perfil = pd.DataFrame(dados, index=RECURSOS)
perfil
```

---
## 2. Mapa de calor

```python
fig, ax = plt.subplots(figsize=(10, 7))
matriz = perfil.values

im = ax.imshow(matriz, cmap="YlGnBu", vmin=0, vmax=3, aspect="auto")

ax.set_xticks(range(len(perfil.columns)))
ax.set_xticklabels(perfil.columns, fontsize=10, weight="bold")
ax.set_yticks(range(len(perfil.index)))
ax.set_yticklabels(perfil.index, fontsize=9.5)
ax.grid(False)

for i in range(matriz.shape[0]):
    for j in range(matriz.shape[1]):
        v = matriz[i, j]
        ax.text(j, i, str(v), ha="center", va="center", fontsize=10,
                color="white" if v >= 2 else "#333333", weight="bold")

cbar = fig.colorbar(im, ax=ax, ticks=[0, 1, 2, 3], shrink=0.75)
cbar.ax.set_yticklabels(["ausente", "possível", "idiomático", "padrão"], fontsize=9)
ax.set_title("Suporte a recursos de paradigma, por linguagem",
             fontsize=14, weight="bold", pad=14)
fig.tight_layout()
plt.show()
```

---
## 3. Quantos paradigmas cabem em cada linguagem?

Somando as notas por linguagem obtemos um índice grosseiro de **amplitude multiparadigma**.


```python
totais = perfil.sum().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(9.5, 4.6))
barras = ax.bar(totais.index, totais.values, color="#1E2761", alpha=0.9)
ax.bar_label(barras, fontsize=10, padding=2)
ax.set_ylabel("Soma das notas de suporte", weight="bold")
ax.set_ylim(0, totais.max() * 1.18)
ax.set_title("Amplitude multiparadigma — soma dos 12 recursos avaliados",
             fontsize=13, weight="bold")
ax.axhline(totais.mean(), ls="--", color="#B85042", lw=1.4)
ax.annotate(f"média = {totais.mean():.1f}", (len(totais) - 0.6, totais.mean() + 0.8),
            color="#B85042", fontsize=9.5, weight="bold")
fig.tight_layout()
plt.show()
```

---
## 4. Quem se parece com quem?

Comparando os perfis dois a dois pela **distância euclidiana**, obtemos uma noção de proximidade
entre linguagens que não depende de sintaxe nem de popularidade — só dos conceitos que cada uma
oferece.


```python
nomes = list(perfil.columns)
M = perfil.values.T.astype(float)
D = np.sqrt(((M[:, None, :] - M[None, :, :]) ** 2).sum(axis=2))

fig, ax = plt.subplots(figsize=(7.5, 6.2))
im = ax.imshow(D, cmap="magma_r")
ax.set_xticks(range(len(nomes))); ax.set_xticklabels(nomes, rotation=45, ha="right", fontsize=10)
ax.set_yticks(range(len(nomes))); ax.set_yticklabels(nomes, fontsize=10)
ax.grid(False)
for i in range(len(nomes)):
    for j in range(len(nomes)):
        ax.text(j, i, f"{D[i, j]:.1f}", ha="center", va="center", fontsize=9,
                color="white" if D[i, j] > D.max() * 0.55 else "#222222")
ax.set_title("Distância conceitual entre linguagens\n(0 = perfis idênticos)",
             fontsize=12.5, weight="bold")
fig.colorbar(im, ax=ax, shrink=0.8, label="distância")
fig.tight_layout()
plt.show()

# o par mais próximo e o mais distante
iu = np.triu_indices(len(nomes), k=1)
pares = sorted(zip(D[iu], zip(np.array(nomes)[iu[0]], np.array(nomes)[iu[1]])))
print("Par mais próximo :", pares[0][1], f"({pares[0][0]:.2f})")
print("Par mais distante:", pares[-1][1], f"({pares[-1][0]:.2f})")
```

### Como ler os gráficos

- O mapa de calor mostra que **não existe uma coluna vazia**: até Go, a linguagem mais enxuta do
  conjunto, tem funções de primeira classe e closures. Multiparadigma é a regra, não a exceção.
- A soma por linguagem premia amplitude, não qualidade. Prolog fica no fim da lista e nem por
  isso é uma linguagem pobre — ela é **profunda em um eixo** que nenhuma outra alcança.
- A matriz de distâncias costuma aproximar Haskell e Elixir, e afastar Prolog de Go. Se a sua
  intuição dizia outra coisa, vale investigar: a discordância é sobre os dados da tabela ou sobre
  o que conta como *parecido*?
